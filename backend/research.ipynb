```python
from pathlib import Path
import json

ROOT_PATH = Path().resolve().parents[0]
DATASET_PATH = ROOT_PATH / "dataset/hr_interview_questions_dataset.json"
print(DATASET_PATH)
```

```python
import ijson

def count_unique(file_path):
    unique_category = set()
    unique_role = set()
    unique_source_type = set()

    with open(file_path, "rb") as file:
        parser = ijson.items(file, "item")
        for item in parser:
            if "category" in item:
                unique_category.add(item['category'])
            if "role" in item:
                unique_role.add(item['role'])
            if "category" in item:
                unique_source_type.add(item['source_type'])
            
    return unique_category, unique_role, unique_source_type
```

```python
categories, roles, source_types = count_unique(DATASET_PATH)
print(f"Уникальные категории: {categories}")
print(f"Уникальные роли: {roles}")
print(f"Уникальные source_types: {source_types}")
```
